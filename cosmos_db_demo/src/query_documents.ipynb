# Query data from cosmos

```python
from generate_documents import connect_cosmos_db_container
from constants import COSMOS_URI, COSMOS_KEY, DB_ID, CONTAINER_ID

container = connect_cosmos_db_container(
    url=COSMOS_URI, key=COSMOS_KEY, db_id=DB_ID, container_id=CONTAINER_ID
)

container
```

```python
import pandas as pd 
items = list(container.query_items(
    query="SELECT * FROM c", enable_cross_partition_query=True
))

df = pd.DataFrame(items)
df.head()
```

```python
for item in items:
    print(item["title"], item["year"])
```

```python
items = list(
    container.query_items(
        query="SELECT * FROM c WHERE c.year < @year",
        parameters=[{"name": "@year", "value": 2015}],
        enable_cross_partition_query=True
    ),
)

for item in items:
    print(item["title"], item["year"])
```
