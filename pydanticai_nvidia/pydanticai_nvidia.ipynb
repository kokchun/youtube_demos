# PydanticAI and Nvidia build models

```python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
import os
from dotenv import load_dotenv

load_dotenv()

model = OpenAIChatModel(
    "nvidia/nemotron-3.5-lightning-30b-a3b",
    provider=OpenAIProvider(
        base_url="https://integrate.api.nvidia.com/v1", api_key=os.environ["NVIDIA_API"]
    ),
)

model
```

```python
agent = Agent(model)

result = await agent.run("What is the capital of Sweden?")
result
```

```python
result.output
```

## Update the model on the fly

```python


new_model = OpenAIChatModel(
    "nvidia/nemotron-3.5-bolt-e4b",
    provider=OpenAIProvider(
        base_url="https://integrate.api.nvidia.com/v1", api_key=os.environ["NVIDIA_API"]
    ),
)

result = await agent.run("Tell me a python joke", model= new_model)
result.output
```

```python
agent = Agent(new_model)

result = await agent.run(
    "Explain how pydanticai differs from other AI frameworks",
    model_settings={"timeout": 60},
)
result.output
```

```python
with open("pydanticai.md", "w") as file:
    file.write(result.output)
```
